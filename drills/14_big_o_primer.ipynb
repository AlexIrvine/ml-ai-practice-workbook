# P14: Big-O primer (lesson + practice)

Every coding interview asks "what's the complexity?". This lesson teaches it from scratch. **No maths beyond counting is needed.**

## Part 1: the lesson

### 1.1 What Big-O means

**Big-O notation** ("big oh", from "**order** of") describes **how the amount of work grows as the input gets bigger**.
- **n** = the size of the input, e.g. how many items are in the list.
- It's **not seconds**. A fast laptop and a slow one give different seconds. Big-O asks: *if the list gets 10× longer, how much more work is there?*

**Everyday example:** you're looking for one name in a pile of n unsorted exam papers.
- Checking the top paper only: **1 step**, however big the pile → **O(1)**, "constant".
- Checking every paper: **n steps** → **O(n)**, "linear". A pile twice as big = twice the work.
- Comparing every paper with every other paper (looking for duplicates): about **n × n steps** → **O(n²)**, "quadratic". Twice as big = **4×** the work.

### 1.2 The five you need (best to worst)

| Big-O | Name | Plain meaning | Typical code | Steps for n = 10 | n = 1,000 | n = 1,000,000 |
|---|---|---|---|---|---|---|
| **O(1)** | constant | same work whatever n is | `nums[0]`, `d[key]`, `x in a_set` | 1 | 1 | 1 |
| **O(log n)** | logarithmic | cut the problem **in half** each step | binary search, `while n > 1: n //= 2` | ~3 | ~10 | ~20 |
| **O(n)** | linear | touch each item **once** (or a few times) | one `for` loop, `sum()`, `max()` | 10 | 1,000 | 1,000,000 |
| **O(n log n)** | "n log n" | sorting | `sorted(nums)`, `nums.sort()` | ~33 | ~10,000 | ~20,000,000 |
| **O(n²)** | quadratic | every item **paired with** every item | a loop inside a loop | 100 | 1,000,000 | 1,000,000,000,000 |

**Read the last column:** with a million items, O(n) is a million steps (well under a second). O(n²) is a **trillion** steps (hours). That's why interviewers care.

(**log n** here = "how many times can you halve n before you reach 1?". 1,000 → 500 → 250 → … → 1 takes about 10 halvings. You learnt logs in maths F4.)

### 1.3 How to work it out: count the loops over the input

| Rule | Code shape | Work | Big-O |
|---|---|---|---|
| 1. No loop over the input | `return nums[0] + nums[-1]` | a fixed few steps | **O(1)** |
| 2. One loop | `for x in nums:` | n | **O(n)** |
| 3. Loops one **after** another **add** | `for x in nums: ...` then `for x in nums: ...` | n + n = 2n | **O(n)** (rule 5) |
| 4. A loop **inside** a loop **multiplies** | `for x in nums:` → `for y in nums:` | n × n | **O(n²)** |
| 5. **Drop constants and smaller terms** | 2n, n + 5, n² + n | | O(n), O(n), O(n²) |

**Why drop them (rule 5)?** Big-O only asks how the work *grows*. 2n and n both double when n doubles, so they're the same "shape". In n² + n, when n = 1,000, n² is 1,000,000 and n is just 1,000, so the n barely matters. Keep the **biggest term only**.

### 1.4 The trap: hidden loops inside built-ins

Some one-liners **secretly loop over the whole list**. This is the most common mistake.

| Operation | Cost | Why |
|---|---|---|
| `len(nums)`, `nums[i]`, `nums.append(x)` | O(1) | Python stores the length; indexing jumps straight there |
| `d[key]`, `key in d`, `x in a_set`, `a_set.add(x)` | **O(1)** | dicts and sets use a **hash**: they jump straight to where the item would be |
| `x in a_list`, `a_list.index(x)`, `a_list.count(x)` | **O(n)** | has to check items one by one |
| `sum(nums)`, `max(nums)`, `min(nums)` | O(n) | looks at every item |
| `nums[:i]` (slicing) | O(i) | copies i items into a new list |
| `sorted(nums)`, `nums.sort()` | O(n log n) | sorting |

**Example:** this looks like one loop, but it's O(n²):
```python
for x in nums:          # n times...
    if x in other_list:  # ...and each `in` checks up to n items
        ...
```
Make `other_list` a **set** first (`other = set(other_list)`) and the `in` becomes O(1), so the whole thing is **O(n)**. This "list → set/dict" trick is behind most of your drills (Two Sum, First Unique Char).

### 1.5 Space complexity: extra memory

**Space** = how much **extra** memory your code creates as n grows. Don't count the input itself.

| You create… | Space |
|---|---|
| a few variables (`total`, `best`, `i`) | **O(1)** |
| a list / dict / set that can hold up to n items (`seen = set()`, a stack, a prefix list) | **O(n)** |
| an n × n table | O(n²) |

**Worked example (your P13 pivot index):** you kept `total` and `running`, just two numbers. So the space is **O(1)**, even for a million-item list. A full `prefix` list would be **O(n)** space.

### 1.6 How to say it in an interview

> "It's **O(n) time** because we go through the list **once**, and each dictionary lookup is O(1). It's **O(n) space** because the dictionary can end up holding **up to n** items."

Template: **"O(__) time because ____; O(__) space because ____."** Always give **both**, and always give the **reason**.

**Where the five show up in ML / AI work:** O(n²) is why naive "compare every document with every other document" doesn't scale in RAG (retrieval-augmented generation). That's why vector databases use clever O(log n)-ish indexes. Sorting a million predictions for a ROC (receiver operating characteristic) curve is O(n log n).

### B1: write `has_duplicate_fast(nums)`

**What you're building:** a function that returns `True` if any value appears more than once in `nums`, otherwise `False`. It must be **O(n)** time. The slow O(n²) version is given below so you can compare.

**Input:**
- `nums`: a list of numbers, e.g. `[3, 8, 1, 8]` (customer IDs, say).

**Why the slow one is O(n²):** it compares every item with every later item: a loop inside a loop.

**The fast idea:** walk through the list **once**, and keep a **set** of the values you've already seen. A set answers "have I seen this?" in O(1).

**Steps, with `nums = [3, 8, 1, 8]`:**

| Step | Current x | Is x in `seen`? | Action | `seen` after |
|---|---|---|---|---|
| start | | | make an empty set: `seen = set()` | {} |
| 1 | 3 | no | add it: `seen.add(3)` | {3} |
| 2 | 8 | no | add it | {3, 8} |
| 3 | 1 | no | add it | {3, 8, 1} |
| 4 | 8 | **yes** | `return True` straight away | |
| (loop ends without a repeat) | | | `return False` | |

**Check:** `has_duplicate_fast([3, 8, 1, 8])` → `True`; `has_duplicate_fast([3, 8, 1])` → `False`; `[]` → `False`.

````python
import time

def time_it(f, n, repeats=3):
    nums = list(range(n))                  # n different numbers: no duplicates (worst case)
    best = float("inf")
    for _ in range(repeats):               # run a few times, keep the fastest (less noise)
        start = time.perf_counter()        # perf_counter() = a precise stopwatch, in seconds
        f(nums)
        best = min(best, time.perf_counter() - start)
    return best

print(f"{'n':>6} | {'slow O(n^2) s':>14} | {'fast O(n) s':>12}")
prev = None
for n in [1000, 2000, 4000]:
    slow, fast = time_it(has_duplicate_slow, n), time_it(has_duplicate_fast, n)
    note = "" if prev is None else f"   slow x{slow / prev[0]:.1f}, fast x{fast / prev[1]:.1f}"
    print(f"{n:>6} | {slow:>14.4f} | {fast:>12.6f}{note}")
    prev = (slow, fast)
````
