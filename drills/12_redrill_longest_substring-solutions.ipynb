## Problem

Given a string `s`, return the **length** of the longest **substring** (contiguous characters) with no repeated characters.

| Input `s` | Output | Why |
|---|---|---|
| `"abcabcbb"` | 3 | `"abc"` |
| `"bbbbb"` | 1 | `"b"` |
| `"pwwkew"` | 3 | `"wke"` (`"pwke"` isn't contiguous) |
| `""` | 0 | empty |

**Constraints:** `0 <= len(s) <= 5 * 10^4`; letters, digits, spaces, symbols; case matters (`"aA"` has no repeats).

**Target:** O(n) time, where n = `len(s)`. Checking every substring is O(n²) or worse.

```python
def length_of_longest_substring(s: str) -> int:
    last_seen = {}          # char -> most recent index
    left = best = 0
    for right, ch in enumerate(s):
        if ch in last_seen and last_seen[ch] >= left:   # repeat INSIDE the window
            left = last_seen[ch] + 1                    # jump past the old copy
        last_seen[ch] = right
        best = max(best, right - left + 1)
    return best
```

```python
# Tests: all asserts must pass silently.
assert length_of_longest_substring("abcabcbb") == 3
assert length_of_longest_substring("bbbbb") == 1          # all duplicates
assert length_of_longest_substring("pwwkew") == 3         # substring, not subsequence
assert length_of_longest_substring("") == 0               # empty
assert length_of_longest_substring("a") == 1              # single char
assert length_of_longest_substring("abcdef") == 6         # no repeats
assert length_of_longest_substring("abba") == 2           # left must never move backwards
assert length_of_longest_substring("a b!a") == 4          # spaces and symbols count
assert length_of_longest_substring("aA") == 2             # case-sensitive
print("all visible tests passed")
```

## Explain it (2–3 lines each)

1. Approach in 1–2 lines. When `ch` has been seen before, where does `left` go, and why there and not "along one"?
2. Time **and** space, biggest term only. Space: what does your dict/set hold at most? (Think about k = the number of different characters possible.)
3. Why does `"abba"` need the check `last_seen[ch] >= left` (or a `while` loop)?

Your answers:

✅ **Model answer**

1. Keep `last_seen[ch] = index`. When `ch` was seen **inside** the current window, jump `left` to `last_seen[ch] + 1`: everything from the old `left` up to the old copy of `ch` contains that repeat, so no valid window can start there. Moving "along one" would just re-check windows that are already known to fail.
2. **O(n) time** (one pass, O(1) dict work per character). **O(min(n, k)) space**, where k = the number of different characters possible: the dict holds at most one entry per distinct character (O(1) for a fixed alphabet).
3. At the final `a` in `"abba"`, `last_seen["a"] = 0`, but `left` is already 2 (after the second `b`). Without the `>= left` check, `left` would jump **back** to 1 and the window `"bba"` (length 3) would be counted, although it contains two `b`s. The check means "only react to repeats inside the current window".
