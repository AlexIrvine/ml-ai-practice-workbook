## Concept primer: sliding window

**What it is:** keep a "window" `[left, right]` over the array and move it forward in **one pass**, updating a running answer, instead of checking every pair/subarray (O(n²)).

**When to use it (signal words):** *contiguous* subarray/substring, "longest/shortest/max/min ... such that", "buy before sell", "best of size k".

**Two flavours:**
- **Fixed size k:** add `a[right]`, drop `a[right-k]`, update answer.
- **Variable size:** expand `right` every step; move `left` forward when the window breaks a rule (or when you find a better start).

**Complexity:** O(n) time (each pointer only moves forward), O(1) space for sums/mins, O(k) if the window keeps a set/dict.

**Tiny worked example (max sum of size k=2 in `[1, 4, 2, 5]`):**
window sums: `1+4=5` → drop 1, add 2 → `6` → drop 4, add 5 → `7`. Answer 7, one pass, no re-summing.
