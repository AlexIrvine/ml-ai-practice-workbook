## Concept primer: stack

**What it is:** a LIFO structure (Python `list` with `.append`/`.pop()` from the end) for matching/undoing in the right order.

**When to use it (signal words):** matching brackets/tags, "next greater element", undo/redo, anything with nested or most-recent-first structure.

**Complexity:** O(n) time, O(n) space (worst case the stack holds everything, e.g. all opens).

**Tiny worked example:** matching `()`: push `(` on open, on `)` pop and check it was `(` — if the stack's empty when you need to pop, or anything's left on the stack at the end, it's invalid.

## Problem: Valid Parentheses

Given a string `s` containing just the characters `(`, `)`, `{`, `}`, `[`, `]`, determine if the input string is valid.

A string is valid if:
1. Open brackets are closed by the same type of bracket.
2. Open brackets are closed in the correct order.
3. Every close has a corresponding open.

**Examples**
```
s = "()"       -> True
s = "()[]{}"    -> True
s = "(]"        -> False
s = "([)]"      -> False   (wrong order)
s = "{[]}"      -> True
```

**Constraints**
- `0 <= len(s) <= 10^4`
- `s` consists only of the characters `()[]{}`

````python
def is_valid(s: str) -> bool:
    pairs = {")": "(", "]": "[", "}": "{"}
    stack = []
    for ch in s:
        if ch in pairs:                                # a closing bracket
            if not stack or stack.pop() != pairs[ch]:  # nothing to match, or wrong type
                return False
        else:
            stack.append(ch)                           # an opening bracket
    return not stack                                   # leftover opens -> invalid
````

````python
# Tests — run this cell. All asserts must pass silently.

assert is_valid("()") == True
assert is_valid("()[]{}") == True
assert is_valid("(]") == False
assert is_valid("([)]") == False
assert is_valid("{[]}") == True
assert is_valid("") == True                  # empty string
assert is_valid("(") == False                 # single unmatched open
assert is_valid(")") == False                 # single unmatched close, nothing to pop
assert is_valid("((((") == False               # all opens, nothing closed
assert is_valid("()(()(()))") == True          # nested + sequential

print("all visible tests passed")
````

## Explain it

In 2-3 lines: approach + time/space complexity. What's the bug people hit when the string starts with a closing bracket?

✅ **Model answer**

Push every opening bracket onto a **stack**; on a closing bracket, pop and check it's the matching type. Valid only if the stack is empty at the end. **O(n) time** (one pass, O(1) push/pop), **O(n) space** (the stack can hold n items, e.g. `"(((("`). The bug: popping from an **empty** stack (`")"`) raises an `IndexError`, so check `if not stack` before popping.
