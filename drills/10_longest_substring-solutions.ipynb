## Pattern note: sliding window, shrinking version

You met sliding window in P4. P4's window never had to shrink: you only tracked the min so far.
This one does: the window has a **rule** ("no repeated characters"). Each step:
1. Move `right` forward by one character.
2. While the window breaks the rule, move `left` forward.
3. The window is valid now, so update the answer with its length.

Both pointers only move forward, so the total work is O(n) even with the inner `while`.

## Problem: Longest Substring Without Repeating Characters

Given a string `s`, return the **length** of the longest **substring** (contiguous characters) that has no repeated characters.

**Examples**
```
s = "abcabcbb"  -> 3    ("abc")
s = "bbbbb"     -> 1    ("b")
s = "pwwkew"    -> 3    ("wke"; "pwke" is not contiguous, so it doesn't count)
s = ""          -> 0
```

**Constraints**
- `0 <= len(s) <= 5 * 10^4`
- `s` can contain letters, digits, spaces and symbols. Case matters: `"aA"` has no repeats.
- Target: **O(n) time**. Checking every substring is O(n²) or worse and will time out.

**Time target: 10 minutes.** Note your start time.

````python
def length_of_longest_substring(s: str) -> int:
    last_seen = {}          # char -> most recent index
    left = best = 0
    for right, ch in enumerate(s):
        if ch in last_seen and last_seen[ch] >= left:   # repeat INSIDE the window
            left = last_seen[ch] + 1                    # jump past the old copy
        last_seen[ch] = right
        best = max(best, right - left + 1)
    return best
````

````python
# Tests: run this cell. All asserts must pass silently.

assert length_of_longest_substring("abcabcbb") == 3
assert length_of_longest_substring("bbbbb") == 1          # all duplicates
assert length_of_longest_substring("pwwkew") == 3         # substring, not subsequence
assert length_of_longest_substring("") == 0               # empty
assert length_of_longest_substring("a") == 1              # single char
assert length_of_longest_substring("abcdef") == 6         # no repeats at all
assert length_of_longest_substring("abba") == 2           # left must never move backwards
assert length_of_longest_substring("a b!a") == 4          # spaces and symbols count: "a b!" / " b!a"
assert length_of_longest_substring("aA") == 2             # case-sensitive

print("all visible tests passed")
````

## Explain it

1. Approach in 1-2 lines.
2. Time AND space, biggest term only. (Space: what does your set/dict hold at most?)
3. Why is the inner `while` loop still O(n) overall, not O(n²)?

✅ **Model answer**

1. Sliding window: grow the window to the right one character at a time; when a character repeats inside the window, move `left` to just past its previous position. Track the longest window seen.
2. **O(n) time; O(min(n, k)) space**, where k = the number of different characters possible (the dict holds at most one entry per distinct character), which is O(1) for a fixed character set.
3. (Set + `while` version.) `left` only ever moves **forwards** and can move at most n times **in total**, across the whole run. So all the inner-loop steps add up to at most n, plus n steps of `right`: 2n → O(n). It's not "n steps for every right".
