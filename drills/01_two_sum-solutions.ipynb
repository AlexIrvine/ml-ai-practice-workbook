## Concept primer: hash map / set

**What it is:** a dict (or set) giving O(1) average-case lookup/insert, trading space for time.

**When to use it (signal words):** "find a pair", "have I seen this before", "count frequency", "does X exist", anything that would otherwise need a nested loop to compare every element against every other element.

**Complexity:** O(n) time, O(n) space — vs. O(n^2) time, O(1) space for the brute-force nested-loop version.

**Tiny worked example:** "contains duplicate" — walk the list once, add each value to a `set`; if a value is already in the set, you've found a duplicate. One pass, no nested loop.

## Problem: Two Sum

Given an array of integers `nums` and an integer `target`, return the **indices** of the two numbers that add up to `target`.

- Exactly one valid answer exists for each input.
- You may not use the same element twice.
- Return the indices in any order.

**Examples**
```
nums = [2, 7, 11, 15], target = 9   -> [0, 1]      (2 + 7 = 9)
nums = [3, 2, 4],      target = 6   -> [1, 2]      (2 + 4 = 6)
nums = [3, 3],         target = 6   -> [0, 1]
```

**Constraints**
- `2 <= len(nums) <= 10^4`
- values can be negative, zero, or duplicated
- there is exactly one valid pair

````python
from typing import List
def two_sum(nums: List[int], target: int) -> List[int]:
    seen = {}                          # value -> index
    for i, x in enumerate(nums):
        need = target - x
        if need in seen:               # O(1) dict lookup
            return [seen[need], i]
        seen[x] = i                    # store AFTER checking, so an element can't pair with itself
    return []
````

````python
# Tests — run this cell. All asserts must pass silently.

def _check(nums, target, expected_sum):
    result = two_sum(nums, target)
    assert isinstance(result, list) and len(result) == 2, f"expected 2 indices, got {result}"
    i, j = result
    assert i != j, "used the same element twice"
    assert nums[i] + nums[j] == expected_sum, f"nums[{i}]+nums[{j}] != {expected_sum}"

_check([2, 7, 11, 15], 9, 9)
_check([3, 2, 4], 6, 6)
_check([3, 3], 6, 6)                      # duplicate values
_check([-1, -2, -3, -4, -5], -8, -8)       # negatives
_check([0, 4, 3, 0], 0, 0)                 # zeros
_check([1, 2, 3, 4, 5], 9, 9)              # target pair not at the ends
_check(list(range(1, 1001)), 1999, 1999)   # larger input, O(n) should still be instant

print("all visible tests passed")
````

## Explain it

In 2-3 lines: what's your approach, and what's the time/space complexity? Why this over brute force?

✅ **Model answer**

One pass with a dictionary of value → index: for each number, check whether `target − x` has already been seen. **O(n) time** (one loop, O(1) dict lookups), **O(n) space** (the dict can hold up to n items). Brute force checks every pair: O(n²) time, which is 100 million pair checks at n = 10⁴.
