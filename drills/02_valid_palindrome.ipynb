## Concept primer: two pointers

**What it is:** walk two indices toward each other (or both forward) instead of nesting loops, exploiting some structure (sorted order, or symmetry) so you never need to re-scan.

**When to use it (signal words):** sorted array + "pair with target", palindrome checks, in-place removal/compaction, merging two sorted sequences.

**Complexity:** O(n) time, O(1) extra space — the whole point versus a nested-loop O(n^2) or an O(n) hash map that still costs O(n) space.

**Tiny worked example:** given a sorted array, find if any pair sums to target — start `lo=0, hi=n-1`; if `nums[lo]+nums[hi] > target` move `hi` left, if `< target` move `lo` right, if equal: found. Never need a hash map because sortedness does the work.
