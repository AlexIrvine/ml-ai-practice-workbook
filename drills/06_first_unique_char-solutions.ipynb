## Concept primer: `collections.Counter`

**What it is:** a dict that counts things for you. `Counter(iterable)` maps each item to how many times it appears.

```python
from collections import Counter
c = Counter("banana")   # Counter({'a': 3, 'n': 2, 'b': 1})
c["a"]                  # 3
c["z"]                  # 0   <- missing keys give 0, no KeyError
c.most_common(1)        # [('a', 3)]
```

**When to use it (signal words):** "frequency", "count", "how many times", "unique", "anagram", "most common".

**Complexity:** building it is O(n); each lookup is O(1) (it's a hash map).

**Same thing by hand:** `counts = {}` then `counts[ch] = counts.get(ch, 0) + 1`. Know both.

## Problem: First Unique Character in a String

Given a string `s`, return the **index** of the first character that appears exactly once. If there isn't one, return `-1`.

**Examples**
```
s = "leetcode"      -> 0   ('l')
s = "loveleetcode"  -> 2   ('v')
s = "aabb"          -> -1
```

**Constraints**
- `0 <= len(s) <= 10^5`
- `s` is lowercase English letters only
- Target: **O(n) time**. Calling `s.count(ch)` for every character is O(n²). Don't.
- **Target time: 8 minutes.**

````python
from collections import Counter

def first_uniq_char(s: str) -> int:
    counts = Counter(s)              # one pass: letter -> how many times
    for i, ch in enumerate(s):       # second pass: first letter with count 1
        if counts[ch] == 1:
            return i
    return -1
````

````python
# Tests: run this cell. All asserts must pass silently.

assert first_uniq_char("leetcode") == 0
assert first_uniq_char("loveleetcode") == 2
assert first_uniq_char("aabb") == -1           # all duplicated
assert first_uniq_char("") == -1               # empty
assert first_uniq_char("z") == 0               # single char
assert first_uniq_char("abcabd") == 2          # 'c' and 'd' both unique, c is first
assert first_uniq_char("aadadaad") == -1

print("all visible tests passed")
````

## Explain it

In 2-3 lines: approach + time/space complexity. Why is space O(1) here even though you build a dict? (Hint: read the constraints.)

✅ **Model answer**

Count every letter with a `Counter` (one pass), then return the index of the first letter whose count is 1 (second pass). **O(n) time** (two passes, 2n → O(n)). **O(1) space**: the dict can hold at most **26 keys** (lowercase English letters only), however long the string is, so its size doesn't grow with n.
