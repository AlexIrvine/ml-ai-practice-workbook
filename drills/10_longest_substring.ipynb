## Pattern note: sliding window, shrinking version

You met sliding window in P4. P4's window never had to shrink: you only tracked the min so far.
This one does: the window has a **rule** ("no repeated characters"). Each step:
1. Move `right` forward by one character.
2. While the window breaks the rule, move `left` forward.
3. The window is valid now, so update the answer with its length.

Both pointers only move forward, so the total work is O(n) even with the inner `while`.
