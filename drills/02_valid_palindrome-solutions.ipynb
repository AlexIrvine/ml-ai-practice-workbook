## Concept primer: two pointers

**What it is:** walk two indices toward each other (or both forward) instead of nesting loops, exploiting some structure (sorted order, or symmetry) so you never need to re-scan.

**When to use it (signal words):** sorted array + "pair with target", palindrome checks, in-place removal/compaction, merging two sorted sequences.

**Complexity:** O(n) time, O(1) extra space — the whole point versus a nested-loop O(n^2) or an O(n) hash map that still costs O(n) space.

**Tiny worked example:** given a sorted array, find if any pair sums to target — start `lo=0, hi=n-1`; if `nums[lo]+nums[hi] > target` move `hi` left, if `< target` move `lo` right, if equal: found. Never need a hash map because sortedness does the work.

## Problem: Valid Palindrome

Given a string `s`, return `True` if it is a palindrome considering **only alphanumeric characters** and **ignoring case**.

**Examples**
```
s = "A man, a plan, a canal: Panama"  -> True
s = "race a car"                      -> False
s = " "                                -> True   (empty after filtering)
```

**Constraints**
- `0 <= len(s) <= 2 * 10^5`
- `s` consists of printable ASCII characters
- Do it in O(n) time. Bonus: O(1) extra space (i.e. don't just build a cleaned copy and reverse it — though get it working first).

````python
def is_palindrome(s: str) -> bool:
    left, right = 0, len(s) - 1
    while left < right:
        if not s[left].isalnum():          # skip punctuation / spaces on the left
            left += 1
        elif not s[right].isalnum():       # ... and on the right
            right -= 1
        elif s[left].lower() != s[right].lower():
            return False
        else:
            left += 1
            right -= 1
    return True

# Simpler O(n)-space version (fine to get working first):
#   cleaned = [c.lower() for c in s if c.isalnum()]
#   return cleaned == cleaned[::-1]
````

````python
# Tests — run this cell. All asserts must pass silently.

assert is_palindrome("A man, a plan, a canal: Panama") == True
assert is_palindrome("race a car") == False
assert is_palindrome("") == True                      # empty string
assert is_palindrome(" ") == True                      # whitespace only -> empty after filtering
assert is_palindrome("a") == True                       # single char
assert is_palindrome("ab") == False                     # two different chars
assert is_palindrome("0P") == False                     # digit vs letter, not equal case-folded
assert is_palindrome("Was it a car or a cat I saw?") == True
assert is_palindrome("12321") == True                   # numeric palindrome

print("all visible tests passed")
````

✅ **Model answer**

Two pointers moving inwards from both ends, skipping non-alphanumeric characters and comparing lower-cased letters. **O(n) time** (each character is visited at most once), **O(1) space** (two indices). The "clean then reverse" version is also O(n) time but **O(n) space**, because it builds a cleaned copy and a reversed copy; the two-pointer loop avoids both copies.
