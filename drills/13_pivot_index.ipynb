## Pattern note: prefix sums (new pattern)

**What:** a **prefix sum** is a running total. `prefix[i]` = sum of the first `i` numbers.

**When to use (signals in the wording):** "sum of a range", "sum to the left / right", "subarray sum", "balance point", "many sum questions on the same list".

**Why:** re-adding a range every time costs O(n) per question. With a running total, any range sum is **one subtraction**, O(1).

**Worked example.** `nums = [1, 7, 3, 6, 5, 6]` (daily £ deposits)

| i | 0 | 1 | 2 | 3 | 4 | 5 | 6 |
|---|---|---|---|---|---|---|---|
| `prefix[i]` (sum of first i) | 0 | 1 | 8 | 11 | 17 | 22 | 28 |

Sum of `nums[2]` to `nums[4]` (3 + 6 + 5 = 14) = `prefix[5] - prefix[2]` = 22 − 8 = **14**. One subtraction, no loop.

You don't always need the whole list: often one **running variable** plus the **total** is enough. That's the key to the problem below.
