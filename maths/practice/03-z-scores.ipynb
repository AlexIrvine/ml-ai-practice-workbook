## Lesson notes (read first)

**Plain English:** a z-score says **how many SDs a value is above or below the mean**. It turns "72 marks" or "£74" into a common unit (SDs), so you can compare things measured on different scales.
- z = 0 → exactly average. z = +2 → 2 SDs above. z = −1 → 1 SD below.
- The sign gives the direction; the size gives how unusual.
- A z-score is **not** a significance test. "z > 0" just means "above average".

**Formula:**  **z = (x − μ) / σ**   (gap from the mean, measured in SDs)
Reverse it to get a value back: **x = μ + z × σ**

**Worked example:** who did better relative to their class?

| | Maths | English |
|---|---|---|
| Your score x | 72 | 72 |
| Class mean μ | 60 | 65 |
| Class SD σ | 8 | 5 |
| Gap x − μ | 12 | 7 |
| z = gap / σ | 12 / 8 = **1.5** | 7 / 5 = **1.4** |

Same raw score, but Maths is slightly more impressive: 1.5 SDs above the class vs 1.4.
A Maths score of 52 → (52 − 60) / 8 = **−1.0** (1 SD below).
Reverse: z = 2 in Maths → x = 60 + 2 × 8 = **76**.

**Standardising a whole column:** compute z for every value. The new column always has **mean 0 and SD 1**. The shape doesn't change, only the units (like converting £ to "SDs").

**Rough guide (only if the data is bell-shaped):** |z| > 2 is unusual, |z| > 3 is very rare. The exact percentages are next lesson (normal distribution, 68–95–99.7).

**Traps:**
1. Subtract the mean **first**, then divide by the SD. 85 / (70 + 10) is not a z-score.
2. On **skewed** data (incomes, transaction amounts) the "|z| > 3 = outlier" rule breaks: the extreme values inflate the mean and SD themselves, so they can hide (Q5 and the code stretch). Use IQR (interquartile range), a log scale, or the median instead.
3. In ML, compute the mean and SD on the **training data only**, then apply them to the test data. Otherwise test information leaks into training.

**ML use:** `StandardScaler` in scikit-learn = z-score every feature, so income (£ thousands) and age (years) are on the same scale. This matters for kNN (k-nearest neighbours), k-means, regularised regression, gradient descent and neural networks. Also used for simple anomaly / fraud flags.

| Code | What it does |
|---|---|
| `(x - x.mean()) / x.std()` | z-scores by hand (NumPy, ÷ n) |
| `scipy.stats.zscore(x)` | same, ddof=0 by default |
| `StandardScaler().fit_transform(X)` | same per column, ddof=0, for ML pipelines |

````python
%matplotlib inline
import numpy as np
from scipy import stats
from sklearn.preprocessing import StandardScaler
````

### New tools (read, run once, see what prints). Example: Maths class

````python
# z by hand: (value - mean) / sd
#    Input: value, mean, sd.  Example: (72 - 60) / 8 -> 1.5
print("z by hand:", (72 - 60) / 8)

# On a whole array, the same line works for every value at once.
#    Example: scores [52, 60, 76] with mean 60, SD 8 -> [-1. 0. 2.]
scores = np.array([52, 60, 76])
print("z of array:", (scores - 60) / 8)

# stats.zscore(array): uses the array's OWN mean and SD (ddof=0).
#    Example: stats.zscore([1, 4, 7]) -> mean 4, SD 2.449 -> [-1.22  0.  1.22]
print("stats.zscore([1,4,7]):", stats.zscore(np.array([1, 4, 7])))

# StandardScaler: the ML version. Needs a 2-D input (rows x columns).
#    .reshape(-1, 1) turns [1, 4, 7] into a single column: [[1], [4], [7]]
#    .fit(train) learns mean & SD from train; .transform(data) applies them.
col = np.array([1, 4, 7]).reshape(-1, 1)
scaler = StandardScaler().fit(col)
print("scaler mean:", scaler.mean_, "scaler SD:", scaler.scale_)
print("scaled:", scaler.transform(col).ravel())   # .ravel() flattens back to 1-D

print("\n--- your answers ---")
````
