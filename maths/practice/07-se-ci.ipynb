**Symbols:**
- μ ("mu") = the true population mean. You never know it; it's what you're estimating.
- x̄ ("x bar") = your sample's mean.
- σ ("sigma") = population SD (standard deviation). s = **sample** SD, your estimate of σ: `np.std(x, ddof=1)`.
- n = sample size.
- **SE = standard error** = σ / √n, in practice s / √n. It's how far a sample mean typically misses μ by.
- **CI = confidence interval**: a range that probably contains μ.
- p̂ ("p hat") = a proportion from your sample, e.g. conversion rate = conversions ÷ visitors.
- 1.96: the z-score with 2.5% above it. Comes from `norm.ppf(0.975)`.

## Lesson notes (read first)

**Plain English:** you only ever see **one** sample, so your x̄ is a guess at μ. Last lesson showed sample means wobble around μ with spread σ / √n. That spread is the **standard error**. A **confidence interval** turns it into an honest answer: "probably between £48 and £56", not just "£52".

**The recipe (95% CI for a mean):**

| Step | Formula | In words |
|---|---|---|
| 1 | x̄ | your best single guess |
| 2 | SE = s / √n | how much that guess typically misses by |
| 3 | margin of error = 1.96 × SE | about 2 SEs either side |
| 4 | **CI = x̄ ± 1.96 × SE** | the range |

Why 1.96? By the CLT, x̄ is ≈ normal around μ with SD = SE. 95% of a normal lies within ±1.96 SDs (the "2" in 68–95–99.7, made exact).

**Worked example 1, average basket value:** a random sample of n = 100 customers, x̄ = £52, s = £20.

| Step | Working | Result |
|---|---|---|
| SE | 20 / √100 = 20 / 10 | £2 |
| margin (95%) | 1.96 × 2 | £3.92 |
| **95% CI** | 52 − 3.92 to 52 + 3.92 | **£48.08 to £55.92** |
| margin (99%) | 2.576 × 2 | £5.15 |
| **99% CI** | 52 ± 5.15 | **£46.85 to £57.15**, wider |

**More confidence → wider interval.** To be surer you've caught μ, you need a bigger net.

**Proportions (conversion rates, accuracy, default rates):** a proportion is a mean of 0s and 1s, so the same recipe works with

**SE = √( p̂ × (1 − p̂) / n )**

**Worked example 2, conversion rate:** 400 visitors, 48 convert. p̂ = 48 / 400 = 0.12.

| Step | Working | Result |
|---|---|---|
| SE | √(0.12 × 0.88 / 400) = √0.000264 | 0.0162 |
| margin | 1.96 × 0.0162 | 0.0318 |
| **95% CI** | 0.12 ± 0.0318 | **8.8% to 15.2%** |

**Worked example 3, model accuracy:** a classifier gets 450 / 500 test examples right, so p̂ = 0.90. SE p̂ √(0.9 × 0.1 / 500) = 0.0134. 95% CI = 0.90 ± 0.026 = **87.4% to 92.6%**. Your "90% accuracy" really means "somewhere around 87–93%".

**What "95% confident" means (a classic interview question):**
- ✅ **Correct:** "If we repeated the sampling many times and built an interval each time, **95% of those intervals would contain the true mean**." The 95% describes the *method*.
- ⚠️ "There's a 95% chance μ is in *this* interval." Everyone says it and it's fine casually. Strictly, μ is fixed, so this one interval either contains it or doesn't.
- ❌ "95% of customers spend between £48 and £56." **No.** The CI is about the *mean*, not individuals. Individual spends vary by s = £20, so they're far more spread out.

**Traps:**
1. SE ≠ SD. SD = spread of **individual values**. SE = spread of **the mean**. SE = SD / √n, which is much smaller.
2. Halving the margin needs **4×** the data (√n again).
3. Small samples (n under ~30): use 2.0–2.3 or more instead of 1.96. That's the **t-distribution**: `scipy.stats.t.ppf(0.975, df=n-1)`. You'll meet it in your A/B drill (P11).
4. Overlapping CIs don't prove "no difference", and gaps between them aren't a formal test. Use a proper hypothesis test (next topic, D4).

**ML / AI use:**
- Reporting model accuracy with ± so you don't oversell 2-point "improvements" (Q5).
- A/B test results: "conversion up 1.2 points, 95% CI −0.3 to +2.7". It includes 0, so it's not convincing.
- Deciding test-set or experiment size: how big n must be for a ±1% margin.
