## Lesson notes (read first)

**Plain English:** lots of real measurements pile up around the average and get rarer the further you go either way: heights, measurement errors, exam marks in a big cohort. Plot them and you get a symmetric **bell curve**. That shape is the **normal distribution**.
- It is fully described by **two numbers: the mean μ (where the centre is) and the SD σ (how wide it is)**.
- It is **symmetric**: mean = median = mode, and the left half mirrors the right.
- Your z-scores from last lesson are how you use it: convert to z, then read off the percentage.

**The 68–95–99.7 rule** (only for normal data):

| Range | % inside | % in **each** tail outside |
|---|---|---|
| μ ± 1σ (z from −1 to 1) | **68%** | 16% |
| μ ± 2σ (z from −2 to 2) | **95%** | 2.5% |
| μ ± 3σ (z from −3 to 3) | **99.7%** | 0.15% |

Slices of one side (handy for odd ranges). The other side mirrors it:

| Slice | % |
|---|---|
| μ to μ + 1σ | 34% |
| μ + 1σ to μ + 2σ | 13.5% |
| μ + 2σ to μ + 3σ | 2.35% |
| beyond μ + 3σ | 0.15% |

**Worked example:** daily ATM withdrawals at a branch ~ N(£8,000, £1,000).

| Question | Convert to z | Answer |
|---|---|---|
| % of days between £7,000 and £9,000 | z = −1 to 1 | **68%** |
| % between £6,000 and £10,000 | z = −2 to 2 | **95%** |
| % of days **above** £10,000 | z > 2 → the tail outside 95%, halved: (100 − 95) / 2 | **2.5%** |
| % **below** £7,000 | z < −1 → (100 − 68) / 2 | **16%** |
| % between £8,000 and £10,000 | z = 0 to 2 → half of 95% | **47.5%** |
| % between £7,000 and £10,000 | z = −1 to 2 → 34 + 47.5 | **81.5%** |

**Method every time:** (1) convert each boundary to z, (2) sketch the bell and shade the region, (3) add up slices or use symmetry.

**For z values that aren't whole numbers** (like 2.5), use code: `scipy.stats.norm.cdf(z)` = % **below** z. `norm.sf(z)` = % **above** z (sf = survival function = 1 − cdf). `norm.ppf(p)` = the reverse: which z has p below it. E.g. `norm.ppf(0.975)` = 1.96, the "1.96" in 95% confidence intervals.

**Traps:**
1. **Not everything is normal.** Incomes, house prices, transaction amounts and waiting times are **right-skewed** (long tail of big values). Counts (complaints per day) are often skewed too. On skewed data the rule is wrong, sometimes badly (code stretch).
2. "Above 2 SDs" is **2.5%**, not 5%. The 5% is split across **both** tails.
3. The rule gives approximate percentages, not exact ones (68.3%, 95.4%, 99.73%).

**ML use:**
- Many models assume normal **noise / errors** (linear regression residuals).
- Neural network **weights start** as small random normal numbers.
- **Anomaly thresholds** ("alert above 3σ") and the cost of false alarms (Q5).
- **Confidence intervals** use 1.96 (next topics: CLT (Central Limit Theorem) → standard error → CIs (confidence intervals)).

### New tools (read, run once, see what prints).
```text
Example: daily ATM withdrawals ~ N(mean 8000, SD 1000)
```

````python
# norm.cdf(x, loc, scale): fraction of values BELOW x.
#    loc = mean, scale = SD. Leave them out and it uses mean 0, SD 1 (z-scores).
#    Example: norm.cdf(9000, 8000, 1000) -> same as norm.cdf(1) -> 0.841 (84.1% below 9000)
print("cdf(9000):", norm.cdf(9000, 8000, 1000), " cdf(z=1):", norm.cdf(1))

# norm.sf(x, loc, scale): fraction ABOVE x (sf = survival function = 1 - cdf).
#    Example: norm.sf(10000, 8000, 1000) -> z = 2 -> 0.0228 (2.3% above)
print("sf(10000):", norm.sf(10000, 8000, 1000))

# Fraction BETWEEN a and b = cdf(b) - cdf(a)   (everything below b, minus everything below a)
#    Example: between 7000 and 9000 -> cdf(9000) - cdf(7000) -> 0.683
print("between 7000 and 9000:", norm.cdf(9000, 8000, 1000) - norm.cdf(7000, 8000, 1000))

# norm.ppf(p): the reverse. Which z has fraction p below it?
#    Example: norm.ppf(0.975) -> 1.96 (97.5% below, so 2.5% above)
print("ppf(0.975):", norm.ppf(0.975))

print("\n--- your answers ---")
````
