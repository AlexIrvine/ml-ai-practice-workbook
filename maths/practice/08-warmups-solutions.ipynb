## Questions

**W1 (median, S1).** Find the median of:    
(a) 14, 3, 9, 21, 6    
(b) 12, 5, 30, 8, 20, 2    
(c) In one line: why do property websites quote the **median** house price, not the mean?

✅ **Model answer**

(a) Sorted 3, 6, **9**, 14, 21 → **9**.
(b) Sorted 2, 5, **8, 12**, 20, 30 → (8 + 12) / 2 = **10**.
(c) House prices are **right-skewed**: a few very expensive homes drag the mean up, so the median is closer to what a typical house costs.

**W2 (derivative = slope, C2).** A model's loss (error) curve is f(x) = x² − 6x. Its derivative (slope formula) is f′(x) = 2x − 6 ("f prime of x").    
(a) Work out f′(1). In plain English, what does that number tell you about the curve at x = 1?    
(b) At what value of x is the minimum (lowest point)? Explain how you found it.    
(c) Gradient descent is standing at x = 1. Should its next step make x **bigger** or **smaller**? Why?

✅ **Model answer**

(a) f′(1) = 2(1) − 6 = **−4**: at x = 1 the curve is going **down**, falling about 4 units of loss for each 1 unit increase in x.
(b) The minimum is where the slope is 0: 2x − 6 = 0 → **x = 3**. (Check: f′(2) = −2 < 0 and f′(4) = +2 > 0, so it really is a low point.)
(c) **Bigger.** Gradient descent steps against the slope: x_new = x − learning rate × f′(x) = 1 − lr × (−4), which increases x, moving downhill towards the minimum at 3.

**W3 (normal tails, D1).** Call-centre wait times are normal with mean (μ, "mu") 50 seconds and standard deviation (σ, "sigma", SD) 10 seconds. Use the 68–95–99.7 rule (no calculator needed).    
(a) What is the z-score for 80 s? What % of calls wait longer than 80 s? Out of 10,000 calls, about **how many calls** is that?    
(b) What is the z-score for 75 s? The % of calls waiting longer than 75 s lies **between which two %s**?    
(c) Same question for 65 s: the % waiting longer lies between which two %s?

✅ **Model answer**

(a) z = (80 − 50) / 10 = **3** → above +3 SD is **0.15%** → 0.0015 × 10,000 = **15 calls**.
(b) z = (75 − 50) / 10 = **2.5** → between the z = 3 tail and the z = 2 tail: **between 0.15% and 2.5%**.
(c) z = (65 − 50) / 10 = **1.5** → between the z = 2 tail (2.5%) and the z = 1 tail (16%): **between 2.5% and 16%**.
