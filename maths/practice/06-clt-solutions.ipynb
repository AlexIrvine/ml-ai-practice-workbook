**Symbols:**
- **Population**: every value there is (e.g. every transaction this year).
- **Sample**: the handful you actually collect.
- n = sample size, the number of values in one sample.
- μ ("mu") = population mean. σ ("sigma") = population SD (standard deviation).
- x̄ ("x bar") = the mean of one sample.
- √n = square root of n (√25 = 5, √100 = 10).

## Lesson notes (read first)

**Plain English:** take a sample, work out its mean, write it down. Do that again and again with fresh samples. You get a pile of sample means. The CLT says that **pile of means** makes a **bell curve (normal)**, even when the original data is not normal at all. The original data keeps its own shape. Only the averages become normal.

**The CLT in three facts** (for random, independent samples with n large enough, rule of thumb **n ≥ 30**):

| Fact | Formula | In words |
|---|---|---|
| 1. Shape | sample means ≈ normal | averages make a bell curve, whatever the data looks like |
| 2. Centre | mean of the x̄'s = μ | averages centre on the true mean (no bias) |
| 3. Spread | SD of the x̄'s = **σ / √n** | bigger samples → averages bunch up tighter |

σ / √n has a name, the **standard error (SE)**. It's next lesson's topic. For now: SE = "how much a sample mean typically misses the true mean by".

**Worked example 1, dice (watch the shape change):** one fair die gives 1–6, each equally likely. That's a **flat** shape, not a bell. μ = 3.5, σ = 1.71.

Average of **2** dice. There are 36 equally likely pairs. Count how many give each average:

| Average | 1 | 1.5 | 2 | 2.5 | 3 | 3.5 | 4 | 4.5 | 5 | 5.5 | 6 |
|---|---|---|---|---|---|---|---|---|---|---|---|
| Ways (out of 36) | 1 | 2 | 3 | 4 | 5 | **6** | 5 | 4 | 3 | 2 | 1 |

An average of 1 needs (1,1), the only way. An average of 3.5 can come from (1,6), (2,5), (3,4), (4,3), (5,2) or (6,1), six ways. The middle wins because **extremes cancel out**. With just 2 dice the flat shape has already become a peak. With 30 dice it's a smooth bell.

Spread of the average, σ / √n:

| n dice averaged | σ / √n | SD of the averages |
|---|---|---|
| 1 | 1.71 / 1 | 1.71 |
| 4 | 1.71 / 2 | 0.85 |
| 25 | 1.71 / 5 | 0.34 |
| 100 | 1.71 / 10 | 0.17 |

Note: **4× the data only halves the spread** (√4 = 2). To halve it again you need 4× more.

**Worked example 2, transactions (a skewed population):** card transactions are **right-skewed**: lots of small ones, a few huge ones. μ = £40, σ = £30. Each day you average a random sample of n = 36.
- Centre of the daily averages = **£40**.
- Spread of the daily averages = 30 / √36 = 30 / 6 = **£5**.
- Daily averages are ≈ normal, so the 68–95–99.7 rule works **on the averages**: 95% of daily averages fall within £40 ± 2 × £5 = **£30 to £50**.
- But **individual** transactions are still skewed. Using the rule on them says 16% fall below μ − σ = £10, and 2.5% below μ − 2σ = **−£20**. A negative transaction is impossible, so the rule is wrong for individual values.

**Why it works (intuition):** each sample mixes some high values and some low ones, and they partly cancel. Extreme averages need *every* value in the sample to be extreme at the same time, and that's rare.

**Traps:**
1. **The CLT is about sample means, not the data.** A bigger sample of house prices is still skewed. Only the *average* of many house prices becomes normal. (This is the most common mistake.)
2. Bigger n makes the means **narrower** (spread σ / √n). It does not change the shape of the population.
3. n ≥ 30 is a rule of thumb. Very skewed data needs more. Fewer than ~30 → use the t-distribution (later).
4. The samples must be **random and independent**. 36 transactions from one customer don't count.
5. √n, not n: doubling n does **not** halve the spread.

**ML / AI use:**
- **A/B tests:** a conversion rate is the **mean of 0s and 1s** (did each visitor buy?). Each visitor is 0 or 1, about as non-normal as data gets, but the rate over thousands of visitors is ≈ normal. That's why z-tests work (Q5).
- **Model accuracy** on a test set is also a mean of 0/1s (right or wrong), so it has a ± margin of error. Next lesson: confidence intervals.
- **Why errors are often normal:** a measurement error is the sum of many small independent nudges. The CLT also applies to sums.
- **Mini-batch training:** the gradient from a batch of 32 is an average, so it's a noisy but roughly normal estimate of the full-data gradient.

**Q1 (easy).** True or false? Give a one-line reason for each.  
(a) "If I collect a big enough sample of house prices, the prices in my sample will be roughly normal."  
(b) "If I take many samples of 50 house prices and work out each sample's mean, those means will be roughly normal."

✅ **Model answer**

(b) **True.** That's the CLT: means of samples of 50 are roughly normal whatever the population's shape, because highs and lows cancel out inside each average.

**Q2 (easy).** A fair die: μ = 3.5, σ = 1.71. You roll 25 dice and average them, and repeat this many times.  
(a) What is the mean of all those averages?  
(b) What is the SD of those averages? Show the formula with numbers.

✅ **Model answer**

(a) **3.5**, the population mean (sample means centre on μ).
(b) SE = σ / √n = 1.71 / √25 = 1.71 / 5 = **0.342**.

**Q3 (medium).** Delivery times are right-skewed: μ = 30 min, σ = 12 min. Each day the manager averages a random sample of 36 deliveries.  
(a) What is the SD of the daily averages?   
(b) Using the 68–95–99.7 rule, 95% of daily averages fall between which two times?  
(c) Can you also say 95% of *individual* deliveries take between 6 and 54 min? Why or why not?

✅ **Model answer**

(a) σ (12 min) is the SD of **individual** deliveries. The SD of the **daily averages** is the standard error: σ / √n = 12 / √36 = 12 / 6 = **2 min**.
(b) The averages are roughly normal (CLT, n = 36), so 95% lie within ±2 SE: 30 ± 2 × 2 → **26 to 34 min**.
(c) **No.** 30 ± 2 × 12 = 6 to 54 uses the 68–95–99.7 rule on individual deliveries, but that rule only holds for **normal** data, and individual delivery times are right-skewed. The CLT makes the **averages** normal, not the individual values.

**Q4 (medium).** You average samples of n = 50 and the spread of your averages is too wide. You want it **half** as wide.  
(a) What sample size do you need?   
(b) Why isn't the answer 100?

✅ **Model answer**

(a) The spread is σ / √n, so to halve it √n must double, i.e. n must go ×4: **n = 200**.
(b) Doubling n to 100 only multiplies √n by √2 ≈ 1.41 (√100 = 10 vs √50 ≈ 7.07), so the spread shrinks by about 29%, not 50%.

**Q5 (applied, interview-style).** An interviewer asks: *"Each visitor either converts or doesn't (1 or 0). That's nowhere near a bell curve. So why can we use a normal-based test to compare conversion rates between version A and version B of a web page?"*  
Answer in 2–4 sentences, as you would out loud. Mention one condition that must hold.

✅ **Model answer**

"A conversion rate **is** a sample mean: it's the average of all those 0s and 1s. The CLT says sample means are approximately normal when n is large, whatever the shape of the individual values, so each group's conversion rate, and the difference between the two, is approximately normal with a standard error we can calculate. That's what the z-test uses. Conditions: visitors must be randomly assigned and independent, and each group needs enough conversions **and** non-conversions (a rule of thumb is at least 10 of each); a very rare event needs a much bigger sample."

````python
%matplotlib inline
import numpy as np
from scipy.stats import skew
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)   # random number generator; 42 = seed, so you get the same numbers each run
````

### New tools (read, run once, see what prints).

````python
# rng.integers(low, high, size): random whole numbers from low up to high - 1 (high is EXCLUDED, like range()).
#    Example: rng.integers(1, 7, size=5) -> 5 die rolls, e.g. [1 5 4 3 3]. 7 is excluded, so the max is 6.
print("5 die rolls:", rng.integers(1, 7, size=5))

# size=(rows, cols) makes a 2-D table instead of a list.
#    Example: rng.integers(1, 7, size=(3, 2)) -> 3 rows, 2 columns = 3 samples of 2 dice each:
#      [[1 4]
#       [6 2]
#       [3 3]]
table = rng.integers(1, 7, size=(3, 2))
print("3 samples of 2 dice:\n", table)

# .mean(axis=1): mean of EACH ROW (one answer per sample). axis=0 would be one per column.
#    Example: [[1 4], [6 2], [3 3]] -> row means [2.5 4.0 3.0], one average per sample.
print("one mean per sample:", table.mean(axis=1))

# rng.exponential(scale, size): right-skewed random numbers with mean = scale and SD = scale.
#    Like transaction amounts: lots of small values, a few big ones.
#    Example: rng.exponential(40, size=5) -> e.g. [12.3  3.1  95.4  27.0  8.8]
print("5 skewed transactions:", rng.exponential(40, size=5).round(1))

# skew(values): measures lopsidedness. 0 = symmetric (like a normal). Above 0 = long right tail.
#    Example: skew([1, 2, 3, 4, 5]) -> 0.0 (symmetric). skew([1, 1, 1, 2, 10]) -> about 1.5 (right tail)
print("skew symmetric:", skew([1, 2, 3, 4, 5]), " skew right tail:", round(skew([1, 1, 1, 2, 10]), 2))

print("\n--- your answers ---")
````

### C1 (checks Q2): a fair die has mean 3.5 and SD 1.71.
```text
    For n = 1, 4 and 25: make 10,000 samples of n dice, average each sample,
    then print the mean and SD of those 10,000 averages.
    Compare the SD with the CLT formula 1.71 / sqrt(n)  (np.sqrt(n)).
```

````python
for n in [1, 4, 25]:
    samples = rng.integers(1, 7, size=(10_000, n))
    means = samples.mean(axis=1)
    formula = 1.71 / np.sqrt(n)
    print("C1 n =", n, " mean of means:", means.mean().round(3), " SD of means:", means.std().round(3), " formula:", round(formula, 3))
# The SD of the means matches 1.71/sqrt(n): 1.71, 0.855, 0.342.
````

### C2 (the big one): a skewed population of 100,000 transactions, mean £40, SD £40.
```text
    (a) Print the population's mean, SD and skew.
    (b) For n = 5, 30 and 100: make 10,000 sample means (use rng.choice below)
        and print their mean, SD and skew.
    Watch: does the skew of the MEANS move towards 0 as n grows? Does the population's?

    rng.choice(values, size=(rows, cols)): picks random values from `values`.
    Example: rng.choice([10, 20, 30], size=(2, 3)) -> e.g. [[20 10 30], [30 30 10]]
```

````python
population = rng.exponential(40, size=100_000)
print('C2 population: mean', population.mean().round(2), ' SD', population.std().round(2), 'skew', round(skew(population), 2))

for n in [5, 30, 100]:
    means = rng.choice(population, size=(10_000, n)).mean(axis=1)
    print('C2 n =', n, ' mean:', means.mean().round(2), ' SD:', means.std().round(2),
          ' formula 40/sqrt(n):', round(40 / np.sqrt(n), 2), ' skew:', round(skew(means), 2))
# The population's skew stays about 2. The skew of the MEANS falls towards 0 as n grows (about 0.9, 0.37, 0.2).
````

### C4 (checks the idea behind Q3): with n = 36 the formula says the SD of the
```text
    means is 40 / 6 = 6.67. So about 95% of sample means should lie within
    40 ± 2 x 6.67. Simulate 10,000 means with n = 36 and print the fraction that do.
```
