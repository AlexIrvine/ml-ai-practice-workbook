## Lesson notes (read first)

**Plain English:** Σ (capital sigma) means "add these up". It's a loop with a running total.

**Worked example:** Σᵢ₌₁³ i² → i = 1, 2, 3 → 1 + 4 + 9 = **14**.
Data x₁ = 20, x₂ = 50, x₃ = 30 → Σ xᵢ = **100**. The little number is an **index** (which item), not a power.

**Mean:** x̄ = (1/n) Σᵢ₌₁ⁿ xᵢ → (1/3) × 100 = **33.33**.

| Symbol | Meaning |
|---|---|
| Σ | add up |
| i | counter (1, 2, 3, …) |
| i = 1 (bottom) / n (top) | start / stop |
| xᵢ | the i-th value |
| x̄ | mean |

**Traps:** Σ xᵢ² = square each, then add. (Σ xᵢ)² = add, then square. Constants come outside: Σ 2xᵢ = 2 Σ xᵢ. **Σ always ends in one number.**

**ML use:** mean, variance (1/n) Σ (xᵢ − x̄)², MSE (1/n) Σ (yᵢ − ŷᵢ)², dot product Σ aᵢbᵢ. In code: `np.sum(...)`.

**Q1.** Σᵢ₌₁³ 2i = 2(1) + 2(2) + 2(3) = 2 + 4 + 6 = **12**.

**Q2.** Σᵢ₌₁⁵ i = 1 + 2 + 3 + 4 + 5 = **15**.

**Q3.** x = 3, 7, 2, 8.
- (a) Σ xᵢ = 3 + 7 + 2 + 8 = **20**
- (b) Σ xᵢ² = 9 + 49 + 4 + 64 = **126** (square each, then add)
- (c) (Σ xᵢ)² = 20² = **400** (add, then square)
- (d) Not the same: 126 ≠ 400. **(b) squares first.** The brackets in (c) say "add first".

**Q4.** Data 1, 3, 5.
- (a) x̄ = (1 + 3 + 5) / 3 = 9 / 3 = **3**
- (b) gaps: 1 − 3 = −2, 3 − 3 = 0, 5 − 3 = +2 → Σ (xᵢ − x̄) = −2 + 0 + 2 = **0**
- (c) It's **always 0**, for any data: the mean is the balance point, so the negative gaps exactly cancel the positive ones. That's why "average gap" can't measure spread. We **square** the gaps first (variance) so they can't cancel.

**Q5.**
- (a) MSE = (1/n) Σᵢ₌₁ⁿ (yᵢ − ŷᵢ)²
- (b) errors: 120 − 110 = 10, 80 − 90 = −10, 100 − 100 = 0 → squared: 100, 100, 0 → sum 200 → MSE = 200 / 3 ≈ **66.7** (units: £²)
- (c) "On average the squared miss is about 67 £²; the square root, about £8.16, is a typical miss per customer, so the predictions are usually within roughly £8." (Saying the root, RMSE = root mean squared error, is clearer for a manager because it's back in £.)

### NumPy in 4 tools (read these, run the file once, see what prints)

```python
# 1. np.array(list): turns a Python list into a NumPy array.
#    Input: a list of numbers.  Example: [20, 50, 30]
spends = np.array([20, 50, 30])
print("spends:", spends)                  # [20 50 30]

# 2. Maths on an array happens to EVERY item at once (no loop needed).
#    Example: spends * 2  ->  [20*2, 50*2, 30*2]  ->  [40 100 60]
print("spends * 2:", spends * 2)
print("spends ** 2:", spends ** 2)        # ** is "to the power": [400 2500 900]
print("spends - 10:", spends - 10)        # [10 40 20]

# 3. np.sum(array): Σ. Adds every item.
#    Example: np.sum([20, 50, 30])  ->  20 + 50 + 30  ->  100
print("np.sum(spends):", np.sum(spends))

# 4. np.mean(array): (1/n) Σ xᵢ. Sum divided by count.
#    Example: np.mean([20, 50, 30])  ->  100 / 3  ->  33.33
#    len(array) gives n, the number of items: len(spends) -> 3
print("np.mean(spends):", np.mean(spends))
print("by hand:", np.sum(spends) / len(spends))

# np.arange(start, stop): numbers from start up to (NOT including) stop.
#    Example: np.arange(1, 4)  ->  [1 2 3]   (stop=4 is excluded)
print("np.arange(1, 4):", np.arange(1, 4))

print("\n--- your answers ---")
```

### Q3: x = 3, 7, 2, 8

### Q4: data 1, 3, 5

```python
d = np.array([1, 3, 5])
mean_d = np.mean(d)
gaps = d - mean_d
print("Q4 mean:", mean_d, "gaps:", gaps, "sum of gaps:", np.sum(gaps) if gaps is not None else None)  # 3.0 [-2 0 2] 0.0
```

### Q5: MSE (mean squared error)

```python
y = np.array([120, 80, 100])       # actual
y_hat = np.array([110, 90, 100])   # predicted
errors = y - y_hat
mse = np.mean(errors ** 2)
print("Q5 errors:", errors, "MSE:", mse)   # [10 -10 0] 66.67
```
