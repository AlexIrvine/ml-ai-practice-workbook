**W1 (algebra, F1).** Simplify −2(3 − x) − (x + 5).  
Is this an expression or an equation? Can you "solve" it?

✅ **Model answer**

−2(3 − x) − (x + 5) = −6 + 2x − x − 5 = **x − 11**. (−2 × −x = **+2x**, and the minus in front of the bracket flips both signs: −x − 5.)
It's an **expression** (no "=" sign), so it can't be solved, only simplified. An **equation** such as x − 11 = 0 can be solved (x = 11).

**W2 (probability, P1).**  
(a) 60% of customers use the app, 50% use the website, 30% use both. What is P(app or website)?  
(b) Each transaction has a 0.1 chance of triggering a fraud alert, independently. What is P(at least one alert in 3 transactions)?

✅ **Model answer**

(a) P(app or web) = 0.6 + 0.5 − 0.3 = **0.8** (subtract the overlap, which was counted twice).
(b) P(no alert in 3) = 0.9³ = 0.729, so P(at least one) = 1 − 0.729 = **0.271**.

**W3 (median, S1).** Find the median of:  
(a) 7, 2, 9, 4, 12  
(b) 3, 8, 1, 10, 6, 15

✅ **Model answer**

(a) Sorted: 2, 4, **7**, 9, 12 → median **7** (the middle of 5 values).
(b) Sorted: 1, 3, **6, 8**, 10, 15 → even count, so average the middle two: (6 + 8) / 2 = **7**. (Not the midpoint of the range, (1 + 15) / 2 = 8.)

**W4 (derivative meaning, C2).** For f(x) = x²:  
(a) What does the derivative at x = 3 tell you, in plain English?  
(b) What is the value of the derivative at the minimum of any smooth curve, and why?

✅ **Model answer**

(a) f′(x) = 2x, so f′(3) = **6**: at x = 3 the curve is going **up**, rising about 6 units of f for each 1 unit of x (its slope there).
(b) **0**: at a minimum the curve is momentarily flat (it has stopped going down and not yet started going up). That's why gradient descent stops when the gradient is 0.

**W5 (bias–variance, M3).** A model scores 99% accuracy on training data and 70% on test data.  
Name the problem (and say high bias or high variance), then give two ways to fix it.

✅ **Model answer**

**Overfitting**, which is **high variance**: the model has memorised the training data (99%) and doesn't generalise (70%). Fixes (any two): a simpler model (fewer features or shallower trees), more or more varied training data, regularisation (L1/L2), early stopping, dropout, or cross-validation to tune complexity.

**W6 (precision vs recall, M1/M2).** 1,000 transactions, 20 of them fraud.  
The model flags 50 as fraud, and 15 of those really are fraud.  
(a) Precision = ? (b) Recall = ? Say in words what each one means here.  
(c) A lazy model predicts "not fraud" for everything. What are its accuracy and recall?

✅ **Model answer**

TP = 15, FP = 50 − 15 = 35, FN = 20 − 15 = 5.
(a) Precision = TP / (TP + FP) = 15 / 50 = **0.30**: of the transactions flagged, 30% really are fraud.
(b) Recall = TP / (TP + FN) = 15 / 20 = **0.75**: the model catches 75% of all real fraud.
(c) All "not fraud": accuracy = 980 / 1,000 = **98%**, recall = 0 / 20 = **0**. High accuracy, useless model, because of the class imbalance.

**W8 (scaling rule, S2).** Salaries have SD (standard deviation) £4,000.  
(a) What is the variance, with units?  
(b) Everyone gets a 1.5× pay rise. New SD? New variance?  
(c) Then everyone gets a flat £2,000 bonus. What happens to SD and variance?  
(d) Which `ddof` (delta degrees of freedom) does `np.std` use by default, and which does pandas `.std()` use?

✅ **Model answer**

(a) Variance = 4,000² = **16,000,000 £²**.
(b) ×1.5: SD → **£6,000**; variance → ×1.5² = ×2.25 → **36,000,000 £²**.
(c) +£2,000 to everyone: **no change** to either (every value shifts equally, so the gaps from the mean are unchanged).
(d) `np.std` defaults to **ddof = 0** (÷ n, population); pandas `.std()` defaults to **ddof = 1** (÷ n − 1, sample).

**W9 (z-scores, S4).**  
(a) Why can one huge outlier stop itself being flagged by the "|z| > 3" rule?  
(b) When standardising features for a model, why fit the scaler on training data only?

✅ **Model answer**

(a) **Masking:** the outlier is part of the data used to compute the mean and SD. It drags the mean towards itself and inflates the SD a lot, so its own z-score shrinks below 3.
(b) To avoid **data leakage:** the test set must stand in for unseen future data. If the scaler's mean and SD include test rows, information from the test set leaks into training and the score is too optimistic. Fit on train, then transform train and test with those numbers.

**W10 (normal distribution, D1).** Delivery times are normal with mean 30 min, SD 4 min.  
(a) What is the z-score for 40 min? Using the 68–95–99.7 rule, the % of deliveries over 40 min lies between which two values?  
(b) Out of 2,000 deliveries, about how many take over 38 min?  
(c) Write the SciPy expression for the fraction within ±1.5 SD of the mean.

✅ **Model answer**

(a) z = (40 − 30) / 4 = **2.5**. It lies between z = 2 (2.5% above) and z = 3 (0.15% above), so the % over 40 min is **between 0.15% and 2.5%**.
(b) 38 min is z = (38 − 30) / 4 = 2 → 2.5% above → 0.025 × 2,000 ≈ **50 deliveries** (exact normal: about 45).
(c) `norm.cdf(1.5) - norm.cdf(-1.5)` (= 0.866), with `from scipy.stats import norm`.
