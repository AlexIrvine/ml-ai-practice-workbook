## Lesson notes (read first)

**Plain English:** the mean is the centre; variance and SD (standard deviation) measure how spread out the data is around it. Square the gaps from the mean (so they can't cancel to 0), then average them.

**Worked example: data 1, 4, 7**

| Step | | Result |
|---|---|---|
| 1. Mean | (1 + 4 + 7) / 3 | **4** |
| 2. Gap from mean | 1−4, 4−4, 7−4 | −3, 0, 3 |
| 3. Square each gap | | 9, 0, 9 |
| 4. Add them (Σ) | 9 + 0 + 9 | **18** |
| 5a. Population variance σ²: ÷ n | 18 / 3 | **6** |
| 5b. Sample variance s²: ÷ (n − 1) | 18 / 2 | **9** |
| 6. SD = √variance | √6, √9 | σ = **2.45**, s = **3** |

**Units:** data in £ → variance in **£²** (meaningless) → SD back in £. Same move as MSE → RMSE.

**Population vs sample:** population (÷ n) = you have every value. Sample (÷ n − 1) = you have some values and estimate the whole; the usual case. The sample mean sits in the middle of its own data, so squared gaps come out too small; ÷ (n − 1) corrects it. Matters for small n, barely for large n.

| Symbol | Meaning |
|---|---|
| μ ("mu") | population mean |
| x̄ ("x-bar") | sample mean |
| σ² = (1/n) Σ (xᵢ − μ)² | population variance |
| σ = √σ² | population SD |
| s² = (1/(n−1)) Σ (xᵢ − x̄)² | sample variance |
| s = √s² | sample SD |

**Code:** `np.std(x)` → ÷ n (ddof=0). `np.std(x, ddof=1)` → ÷ (n − 1). ddof = *delta degrees of freedom* = the number subtracted from n. **pandas `.std()` defaults to ddof=1.**

**ML use:** `StandardScaler` (subtract mean, divide by SD), z-scores and outliers, standard errors and confidence intervals (A/B tests).

### New tools (read, run once, see what prints). Example data: 1, 4, 7

````python
ex = np.array([1, 4, 7])

# np.sqrt(number_or_array): square root.
#    Input: a number or array.  Example: np.sqrt(9) -> 3.0
print("np.sqrt(9):", np.sqrt(9))

# np.var(array, ddof=0): variance. ddof = number subtracted from n.
#    ddof=0 -> divide by n (population).  ddof=1 -> divide by n-1 (sample).
#    Example: np.var([1, 4, 7])         -> 18 / 3 -> 6.0
#             np.var([1, 4, 7], ddof=1) -> 18 / 2 -> 9.0
print("np.var ddof=0:", np.var(ex), " ddof=1:", np.var(ex, ddof=1))

# np.std(array, ddof=0): standard deviation = sqrt of variance. Same ddof rule.
#    Example: np.std([1, 4, 7]) -> sqrt(6) -> 2.449
print("np.std ddof=0:", np.std(ex), " ddof=1:", np.std(ex, ddof=1))

# pd.Series(list).std(): pandas SD. DEFAULT IS ddof=1 (sample)!
#    Example: pd.Series([1, 4, 7]).std() -> 3.0  (not 2.449)
print("pandas .std():", pd.Series(ex).std())

print("\n--- your answers ---")
````

### Q1 + Q2: data 5, 7, 9, 11, 13. Do it BY HAND in code first, then check.

````python
x = np.array([5, 7, 9, 11, 13])
n = len(x)
mean = np.sum(x) / n
gaps = x - mean
sq = gaps ** 2
pop_var = np.sum(sq) / n
sample_var = np.sum(sq) / (n - 1)
print('Q1 by hand: 9', mean, '8', pop_var, '2.83', np.sqrt(pop_var))
print('Q2 by hand: 10', sample_var, '3.16', np.sqrt(sample_var))
print('Q1/Q2 check with np.var / np.std:', np.var(x), np.std(x), np.var(x, ddof=1), np.std(x, ddof=1))
````

### Q5: NumPy vs pandas on 6 spends

````python
spends = [30, 45, 50, 60, 80, 95]
print('Q5 np.std:', np.std(spends))                          # 21.79 (divides by n)
print('Q5 pandas .std():', pd.Series(spends).std())          # 23.87 (divides by n - 1)
print('Q5 np.std with ddof=1:', np.std(spends, ddof=1))      # 23.87
````

### STRETCH: prove n - 1 works, by simulation.
```text
A population whose true variance we KNOW (= 100, i.e. SD 10).
Draw 10,000 small samples (n = 3). For each, compute the variance
both ways. Average them. Which average is closer to the true 100?
```

````python
rng = np.random.default_rng(0)                    # random number generator, seed 0 so results repeat
samples = rng.normal(loc=50, scale=10, size=(10_000, 3))   # 10,000 rows, each a sample of 3

var_n = np.var(samples, axis=1, ddof=0)
var_n_minus_1 = np.var(samples, axis=1, ddof=1)
print('STRETCH avg var (divide by n):', var_n.mean(), ' avg var (divide by n-1):', var_n_minus_1.mean())
print("True variance: 100")
# ÷ n averages about 66.7 (= 100 × 2/3, too small); ÷ (n - 1) averages about 100. n - 1 is unbiased.
````
